# Tokenizer 基本使用

```python
from transformers import AutoTokenizer
```

```python
sen = "弱小的我也有大梦想!"
```

## Step1 加载与保存

```python
# 从 HuggingFace 加载, 输入模型名称, 即可加载对于的分词器
tokenizer = AutoTokenizer.from_pretrained("uer/roberta-base-finetuned-dianping-chinese")
tokenizer
```

```python
# tokenizer 保存到本地
tokenizer.save_pretrained("./roberta_tokenizer")
```

```python
!ls -al ./roberta_tokenizer
```

```python
# 从本地加载 tokenizer
tokenizer = AutoTokenizer.from_pretrained("./roberta_tokenizer/")
tokenizer
```

## Step2 句子分词

```python
tokens = tokenizer.tokenize(sen)
tokens
```

```python
len(tokens)
```

## Step3 查看词典

```python
tokenizer.vocab
```

```python
tokenizer.vocab_size
```

## Step4 索引转换

```python
# 将词序列转换为 id 序列
ids = tokenizer.convert_tokens_to_ids(tokens)
ids
```

```python
len(ids)
```

```python
# 将 id 序列转换为 token 序列
tokens = tokenizer.convert_ids_to_tokens(ids)
tokens
```

```python
# 将 token 序列转换为 string
str_sen = tokenizer.convert_tokens_to_string(tokens)
str_sen
```

###  更便捷的实现方式

```python
# 将字符串转换为 id 序列, 又称之为编码
ids = tokenizer.encode(sen)
ids
```

```python
ids = tokenizer.encode(sen, add_special_tokens=False)
ids
```

```python
ids = tokenizer.encode(sen, add_special_tokens=True)
# 等价于 ids = tokenizer.encode(sen)
ids
```

```python
# 将 id 序列转换为字符串, 又称之为解码
str_sen = tokenizer.decode(ids, skip_special_tokens=False)
str_sen
```

```python
str_sen = tokenizer.decode(ids, skip_special_tokens=True)
str_sen
```

## Step5 填充与截断

```python
# 填充
ids = tokenizer.encode(sen, padding="max_length", max_length=15)
ids
```

```python
# 截断
ids = tokenizer.encode(sen, max_length=5, truncation=True)
ids
```

## Step6 其他输入部分

```python
ids = tokenizer.encode(sen, padding="max_length", max_length=15)
ids
```

```python
attention_mask = [1 if idx != 0 else 0 for idx in ids]
attention_mask
```

```python
token_type_ids = [0] * len(ids)
token_type_ids
```

```python
ids, attention_mask, token_type_ids
```

## Step7 快速调用方式

```python
inputs = tokenizer.encode_plus(sen, padding="max_length", max_length=15)
inputs
```

```python
inputs = tokenizer(sen, padding="max_length", max_length=15)
inputs
```

## Step8 处理 batch 数据

```python
sens = ["弱小的我也有大梦想", "有梦想谁都了不起", "追逐梦想的心，比梦想本身，更可贵"]
res = tokenizer(sens)
res
```

```python
%%time
# 单条循环处理
for i in range(1000):
    tokenizer(sen)
```

```python
%%time
# 处理 batch 数据
res = tokenizer([sen] * 1000)
```

```python
tokenizer
```

# Fast / Slow Tokenizer

```python
sen = "弱小的我也有大Dreaming!"
```

```python
fast_tokenizer = AutoTokenizer.from_pretrained(
    "uer/roberta-base-finetuned-dianping-chinese"
)
fast_tokenizer
```

```python
slow_tokenizer = AutoTokenizer.from_pretrained(
    "uer/roberta-base-finetuned-dianping-chinese", use_fast=False
)
slow_tokenizer
```

```python
%%time
# 单条循环处理
for i in range(10000):
    fast_tokenizer(sen)
```

```python
%%time
# 单条循环处理
for i in range(10000):
    slow_tokenizer(sen)
```

```python
%%time
# 处理 batch 数据
res = fast_tokenizer([sen] * 10000)
```

```python
%%time
# 处理 batch 数据
res = slow_tokenizer([sen] * 10000)
```

```python
inputs = fast_tokenizer(sen, return_offsets_mapping=False)
inputs
```

```python
inputs = fast_tokenizer(sen, return_offsets_mapping=True)
inputs
```

```python
inputs.word_ids()
```

```python
inputs = slow_tokenizer(sen, return_offsets_mapping=False)
inputs
```
