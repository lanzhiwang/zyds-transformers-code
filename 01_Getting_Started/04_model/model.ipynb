# 模型调用

```python
sen = "弱小的我也有大梦想！"
tokenizer = AutoTokenizer.from_pretrained("hfl/rbt3")
inputs = tokenizer(sen, return_tensors="pt")
inputs
```

## 不带 Model Head 的模型调用

```python
# 输出中没有 attentions
model = AutoModel.from_pretrained("hfl/rbt3", output_attentions=False)
```

```python
output = model(**inputs)
output
```

```python
# 输出中有 attentions
model = AutoModel.from_pretrained("hfl/rbt3", output_attentions=True)
```

```python
output = model(**inputs)
output
```

```python
output.last_hidden_state.size()
```

```python
len(inputs["input_ids"][0])
```

## 带 Model Head 的模型调用

```python
from transformers import AutoModelForSequenceClassification
```

```python
# 二分类任务
clz_model = AutoModelForSequenceClassification.from_pretrained("hfl/rbt3")
```

```python
clz_model(**inputs)
```

```python
clz_model.config.num_labels
```

```python
# 十分类任务
clz_model = AutoModelForSequenceClassification.from_pretrained(
    "hfl/rbt3", num_labels=10
)
```

```python
clz_model(**inputs)
```

```python
clz_model.config.num_labels
```
